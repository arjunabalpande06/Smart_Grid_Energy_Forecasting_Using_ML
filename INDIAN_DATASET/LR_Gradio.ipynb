### Environment Setup

```python
# Local environment setup
from pathlib import Path
```

Librairies

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from sklearn.model_selection import train_test_split
import joblib
import gradio as gr

print("Libraries imported successfully!")
```

Load Dataset

```python
from pathlib import Path

DATASET_DIR = Path("dataset")
if not DATASET_DIR.exists() and (Path("INDIAN_DATASET") / "dataset").exists():
    DATASET_DIR = Path("INDIAN_DATASET") / "dataset"

file_path = DATASET_DIR / "hourlyLoadDataIndia.xlsx"

df = pd.read_excel(file_path)

df
```

```python
print("Dataset loaded successfully!")
print("Shape:", df.shape)

display(df.head())
```

Check and clean

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')

df = df.dropna(subset=['datetime'])

df = df.sort_values('datetime').reset_index(drop=True)

print("Final dataset shape:", df.shape)
display(df.head())
```

Create Time based features

```python
df['hour'] = df['datetime'].dt.hour
df['day'] = df['datetime'].dt.day
df['day_of_week'] = df['datetime'].dt.dayofweek
df['month'] = df['datetime'].dt.month
df['year'] = df['datetime'].dt.year
df['day_of_year'] = df['datetime'].dt.dayofyear

print(df[['datetime', 'hour', 'day', 'day_of_week',
          'month', 'year', 'day_of_year']].head())
```

Select Target

```python
target = 'National Hourly Demand'

print("Target:", target)
print("Target unit: MW")
```

```python
print(df[target].describe())
```

Select input and output

```python
features = [
    'hour',
    'day',
    'day_of_week',
    'month',
    'year',
    'day_of_year'
]

X = df[features]
y = df[target]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

Train test split

```python
split_index = int(len(df) * 0.80)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

Training Linear Regression

```python
model = LinearRegression()

model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
```

Make prediction

```python
y_pred = model.predict(X_test)

print("Predictions generated successfully!")
print(y_pred[:10])
```

Model Evaluation

```python
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print("LINEAR REGRESSION PERFORMANCE")
print("--------------------------------")
print("MAE  :", round(mae, 4))
print("MSE  :", round(mse, 4))
print("RMSE :", round(rmse, 4))
print("R2   :", round(r2, 4))
```

Graph

```python
plt.figure(figsize=(15, 6))

plt.plot(y_test.values[:500], label='Actual Demand')
plt.plot(y_pred[:500], label='Predicted Demand')

plt.title('Linear Regression - Actual vs Predicted National Electricity Demand')
plt.xlabel('Time')
plt.ylabel('Electricity Demand (MW)')
plt.legend()
plt.grid(True)

plt.show()
```

Save Model

```python
from pathlib import Path

MODEL_DIR = Path("Linear Regression Model Data")
if not MODEL_DIR.exists() and (Path("INDIAN_DATASET") / "Linear Regression Model Data").exists():
    MODEL_DIR = Path("INDIAN_DATASET") / "Linear Regression Model Data"
MODEL_DIR.mkdir(parents=True, exist_ok=True)

model_path = MODEL_DIR / "linear_regression_model.pkl"

joblib.dump(model, model_path)

print(f"Model saved successfully to {model_path}!")
```

```python
def predict_load(date, time):
    try:
        datetime_value = pd.to_datetime(f"{date} {time}")

        input_data = pd.DataFrame({
            'hour': [datetime_value.hour],
            'day': [datetime_value.day],
            'day_of_week': [datetime_value.dayofweek],
            'month': [datetime_value.month],
            'year': [datetime_value.year],
            'day_of_year': [datetime_value.dayofyear]
        })

        prediction = model.predict(input_data)[0]

        return f"{prediction:,.2f} MW"

    except Exception as e:
        return f"Error: {str(e)}"
```
